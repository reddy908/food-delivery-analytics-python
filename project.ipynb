```python
df["Delivery_Time_Min"].describe()
```

```python
q1=df["Delivery_Time_Min"].quantile(0.25)
q3=df["Delivery_Time_Min"].quantile(0.75)
```

```python
iqr=q3-q1
low=q1-1.5*iqr
high=q3+1.5*iqr
df[(df["Delivery_Time_Min"]<low) | (df["Delivery_Time_Min"]>high)]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
df["Delivery_Time_Min"].plot(kind="box")
plt.title("outliers detection in delivery_time")
plt.show()
```

```python
outliers=df[(df["Delivery_Time_Min"]<low) | (df["Delivery_Time_Min"]>high)]
outliers[[
       'Order_Value', 'Distance_KM', 'Preparation_Time_Min',
        'Order_Hour', 'Delivery_Time_Min', 'Customer_Rating'
       ]].sort_values(by='Delivery_Time_Min')
```

```python
#delayed_avg
delayed_avg=df[df["Order_Status"]== "Delayed"].groupby("Restaurant")["Delivery_Time_Min"].mean().sort_values()
delayed_avg
```
